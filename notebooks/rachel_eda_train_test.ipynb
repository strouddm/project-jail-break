```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Multi-turn data

X_test_m = pd.read_csv('../data/processed/multiturn_X_test.csv')
y_test_m = pd.read_csv('../data/processed/multiturn_Y_test.csv')

print(X_test_m.shape)
print(y_test_m.shape)
```

```python
# Viewing sample data

print(X_test_m[['conversation', 'conversation_proc']])
print(y_test_m)
```

```python
print('Harm distribution Train (Multi):')
print(y_test_m['harm'].value_counts() / len(y_test_m))
```

```python
# Single-turn data

X_train_s = pd.read_csv('../data/processed/singleturn_X_train.csv')
y_train_s = pd.read_csv('../data/processed/singleturn_Y_train.csv')
X_test_s = pd.read_csv('../data/processed/singleturn_X_test.csv')
y_test_s = pd.read_csv('../data/processed/singleturn_Y_test.csv')
X_val_s = pd.read_csv('../data/processed/singleturn_X_val.csv')
y_val_s = pd.read_csv('../data/processed/singleturn_Y_val.csv')

print(X_train_s.shape)
print(y_train_s.shape)
print(X_test_s.shape)
print(y_test_s.shape)
print(X_val_s.shape)
print(y_val_s.shape)
```

```python
# Viewing sample data

print(X_train_s[['conversation', 'conversation_proc']])
print(y_train_s)
```

```python
print('Harm distribution Train (Single):')
print(y_train_s['harm'].value_counts() / len(y_train_s))

print('\nHarm distribution Validation (Single):')
print(y_val_s['harm'].value_counts() / len(y_val_s))

print('\nHarm distribution Test (Single):')
print(y_test_s['harm'].value_counts() / len(y_test_s))
```
